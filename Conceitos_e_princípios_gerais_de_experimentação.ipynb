#Conceitos e princípios gerais de experimentação - Em Construção

## 1. O que é experimentação?

A experimentação pode ser entendida como uma forma planejada e organizada de investigar um problema. Em vez de apenas observar um fenômeno e tirar conclusões imediatas, o pesquisador define previamente como os dados serão obtidos, quais condições serão comparadas e de que maneira os resultados serão analisados.

Por isso, o processo de experimentação não inicia no momento da coleta de dados. Ele começa antes, com a definição clara do que que pretende descobrir. O pesquisador identifica os fatores que serão estudados, os tratamentos que serão comparados, as unidades experimentais nas quais esses tratamentos serão aplicados e a variável resposta que será medida para verificar os possíveis efeitos.

Também é necessário considerar que as unidades experimentais podem apresentar diferenças naturais entre si. Mesmo quando recebem o mesmo tratamento, elas podem apresentar resultados diferentes devido às suas características próprias ou a condições que não foram totalmente controladas. Para lidar com essa variabilidade, são utilizadas estratégias como a casualização e a repetição, que ajudam a tornar as comparações mais confiáveis.

Em um experimento clássico, o pesquisador modifica deliberadamente uma ou mais condições e observa os efeitos produzidos sobre a variável resposta. Quanto maior for o controle sobre os demais fatores que podem interferir no resultado, maior será a a possibilidade de atribuir as variações observadas ao fator que está sendo estudado.

Entretanto, nem sempre é possível controlar todas as condições envolvidas. Nesses casos, o planejamento experimental e as técnicas estatísticas ajudam a separar, na medida do possível, o efeito de interesse das variações externas ou aleatórias. Assim, a experimentação não se resume à aplicação de tratamentos, mas envolve todo o planejamento necessário para produzir resultados válidos e interpretá-los de forma adequada.

## 2. Elementos que formam um experimento

### 2.1 Tratamento

O tratamento é a condição que será aplicada ou comparada durante o experimento. Ele representa aquilo que o pesquisador deseja avaliar.

Os tratamentos podem ser quantitativos, quando seus níveis podem ser colocados em uma escala numérica, como diferentes doses, temperaturas ou tempos. Também podem ser qualitativos, quando representam categorias, como marcas, métodos ou variedades.

### 2.2 Fator e níveis

O fator é a variável que está sendo estudada. Os níveis são as formas ou valores que esse fator pode assumir.

Por exemplo, se eu quiser avaliar o efeito de três ligas metálicas sobre a dilatação de uma peça, o fator será o tipo de liga e os três tipos de liga serão os níveis. A dilatação será a resposta observada.

### 2.3 Unidade experimental

A unidade experimental é o local, objeto ou indivíduo que recebe o tratamento de forma independente.

Essa parte exige atenção porque a unidade experimental não é necessariamente cada medição realizada. Posso medir várias vezes a mesma peça, mas isso não significa que possuo várias unidades experimentais. As medições pertencem à mesma unidade.

### 2.4 Variável resposta

A variável resposta é aquilo que será medido para verificar se houve efeito dos tratamentos aplicados.

Ela pode ser discreta, quando representa uma contagem, ou contínua, quando pode assumir valores dentro de um intervalo, como peso, área, temperatura ou comprimento.

### 2.5 Erro experimental

O erro experimental é a variação natural e ao acaso que aparece entre unidades submetidas ao mesmo tratamento.

Esse erro não significa necessariamente que alguém executou o experimento de maneira errada. Ele representa diferenças naturais ou fatores que não puderam ser totalmente controlados, como pequenas variações do ambiente, do material ou do processo de medição.

Por isso, o planejamento precisa permitir que essa variabilidade seja estimada e considerada na análise.
